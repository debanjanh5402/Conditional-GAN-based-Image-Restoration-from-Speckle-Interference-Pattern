### Train, Test Split

```python
import os
import random
import shutil

train_dir = '/Users/debanjan_5402/Desktop/Conditional-GAN-based-Image-Restoration-from-Speckle-Interference-Pattern/dataset/train'
test_dir = '/Users/debanjan_5402/Desktop/Conditional-GAN-based-Image-Restoration-from-Speckle-Interference-Pattern/dataset/test'

all_folders = [f for f in os.listdir(train_dir) if os.path.isdir(os.path.join(train_dir, f))]
print(f"Total available folders: {len(all_folders)}")

test_folders = random.sample(all_folders, 194)

# 8. Move folders to the Test directory
for folder in test_folders:
    src_path = os.path.join(train_dir, folder)
    dst_path = os.path.join(test_dir, folder)
    shutil.move(src_path, dst_path)

print(f"Successfully moved {len(test_folders)} folders to test.")
```

### Random sampling Sample dataset

```python
import os
import random
import shutil
```

```python
src_dir = '/Users/debanjan_5402/Desktop/Conditional-GAN-based-Image-Restoration-from-Speckle-Interference-Pattern/dataset/train'
dst_dir = '/Users/debanjan_5402/Desktop/Conditional-GAN-based-Image-Restoration-from-Speckle-Interference-Pattern/_sample_dataset/train'

all_folders = [f for f in os.listdir(src_dir) if os.path.isdir(os.path.join(src_dir, f))]
selected_folders = random.sample(all_folders, 16)

for folder in selected_folders:
    src_path = os.path.join(src_dir, folder)
    dst_path = os.path.join(dst_dir, folder)
    shutil.copytree(src_path, dst_path)
```
